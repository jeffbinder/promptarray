```python
import torch
from generator import PromptArrayGenerator
from transformers import AutoTokenizer, AutoModelForCausalLM

import os
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"

model_name_or_path = 'openai/gpt-oss-20b'
device = 'cuda' if torch.cuda.is_available() else 'mps' if torch.backends.mps.is_available() else 'cpu'

# Initialize the model and tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name_or_path)
model = AutoModelForCausalLM.from_pretrained(
    model_name_or_path,
    device_map='cpu' if device == 'mps' else 'auto',
    dtype=torch.bfloat16
)
if device == 'mps':
    model = model.to('mps')
model.eval()

# Initialize PromptArray
generator = PromptArrayGenerator(
    model,
    tokenizer
)
```

```python
length = 2000
do_sample = True
temperature = 0.3
k = 10
p = 1.0
repetition_penalty = 2.0
bad_words = []
num_return_sequences = 2
overlap_factor = 0.25
chat_mode = True
chat_mode_think_first = True
chat_mode_max_thought_length = 2000

prompt_text = '''Generate {a mellifluous and euphuistic literary description of&an ungrammatical and badly spelled reddit shitpost about} what it's like to live in New York. Keep to one paragraph.'''
#prompt_text = '''Write a short, one-paragraph description of a {snake&{cat/kitty}}.'''

import time
start_time = time.time()
outputs = generator(
    prompt=prompt_text,
    chat_mode=chat_mode,
    chat_mode_think_first=chat_mode_think_first,
    chat_mode_max_thought_length=chat_mode_max_thought_length,
    num_return_sequences=num_return_sequences,
    max_length=length,
    do_sample=do_sample,
    temperature=temperature,
    top_k=k,
    top_p=p,
    repetition_penalty=repetition_penalty,
    bad_words=bad_words,
    overlap_factor=overlap_factor,
    verbose=True
)
print(f"Time: {time.time() - start_time}s")

for i, output in enumerate(outputs):
    if num_return_sequences > 1:
        print(f'Generated sequence {i}:')
    print(output)

```

```python
length = 2000
do_sample = True
temperature = 0.3
k = 10
p = 1.0
repetition_penalty = 2.0
bad_words = []
num_return_sequences = 1
num_runs = 25
overlap_factor = 0.25
chat_mode = True
chat_mode_think_first = True
chat_mode_max_thought_length = 2000

prompt_text = '''Write instructions for how to boil an egg, without using any articles (a, an, the). Output only the instructions, no commentary or other text.'''
#prompt_text = '''Write a short, one-paragraph description of a {snake&{cat/kitty}}.'''

import time
start_time = time.time()
outputs = []
for _ in range(num_runs):
    outputs += generator(
        prompt=prompt_text,
        chat_mode=chat_mode,
        chat_mode_think_first=chat_mode_think_first,
        chat_mode_max_thought_length=chat_mode_max_thought_length,
        num_return_sequences=num_return_sequences,
        max_length=length,
        do_sample=do_sample,
        temperature=temperature,
        top_k=k,
        top_p=p,
        repetition_penalty=repetition_penalty,
        bad_words=bad_words,
        overlap_factor=overlap_factor,
        verbose=True
    )
print(f"Time: {time.time() - start_time}s")

for i, output in enumerate(outputs):
    if num_return_sequences > 1:
        print(f'Generated sequence {i}:')
    print(output)

import re
article_pattern = re.compile(r'\b(a|an|the)\b', re.IGNORECASE)
count_with_articles = sum(1 for output in outputs if article_pattern.search(output))
print(f"Number of outputs containing 'a', 'an', or 'the': {count_with_articles}")
```
